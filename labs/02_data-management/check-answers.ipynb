```python
import json
import sys
import re
from pathlib import Path
```

```python
with open("data-format.ipynb", 'r', encoding='utf-8') as f:
    notebook = json.load(f)

with open("data-format-answers.ipynb", 'r', encoding='utf-8') as f:
    notebook_answers = json.load(f)
```

```python
output_lines = []
source_lines = []
for i, cell in enumerate(notebook.get('cells', []), 1):
    # Extract cell content
    cell_type = cell.get('cell_type', [])
    if cell_type == 'code':
        cell_output = cell.get('outputs', [])
        output_lines.append(cell_output)  
        cell_source = cell.get('source', [])
        source_lines.append(cell_source)  
```

```python
output_lines_answers = []
source_lines_answers = []
for i, cell in enumerate(notebook_answers.get('cells', []), 1):
    # Extract cell content
    cell_type = cell.get('cell_type', [])
    if cell_type == 'code':
        cell_output = cell.get('outputs', [])
        output_lines_answers.append(cell_output) 
        cell_source = cell.get('source', [])
        source_lines_answers.append(cell_source)  
```
