```python

'''
# data extracted from: https://doi.org/10.5281/zenodo.23046811
'''

from pandas import read_csv
from datetime import datetime
import matplotlib.pyplot as plt
from labs01utils import download_file_from_github

repository = "GEOS-Chem-IGE/hough2026_atmo-plast-emis_harmonize"
commit = "a0ba5c3a6a5e0271ef9860ea8a5f8b73fce0e44d"
filepath = "data/processed/obs_evangelou2026-revised.csv"

download_file_from_github(repository, commit, filepath)

data = read_csv("obs_evangelou2026-revised.csv")

year_concentration = data.year[data.measure == "concentration"]
number_concentration = data.number[data.measure == "concentration"]

year_deposition = data.year[data.measure == "deposition"]
number_deposition = data.number[data.measure == "deposition"]

plt.plot(year_concentration, number_concentration, 'ko')
# Plot a time series of Grenoble's population (with circles)
ax = plt.gca()
ax.set_yscale('log')
# Calculate the average population in Grenoble since 1990
plt.show()
```

```python
plt.plot(year_deposition, number_deposition, 'ko')
# Plot a time series of Grenoble's population (with circles)
ax = plt.gca()
ax.set_yscale('log')
# Calculate the average population in Grenoble since 1990
plt.show()
```

```python
conc_avg = number_concentration.mean()
dep_avg = number_deposition.mean()
print(f"Average atmospheric microplastics composition = {conc_avg:.4f} particles/m3")
print(f"Average microplastics deposition rate = {dep_avg:.4f} particles/m2/d")
```
