```python
# Analyze the observation data from Evangelou et al. (2026), DOI: 10.5281/zenodo.23046811 
# used in Reduced global atmospheric microplastic emissions from size-harmonized observations" (Hough et al., 2026)

from pandas import read_csv
from matplotlib.pyplot import plot

data = read_csv("obs_evangelou2026-revised.csv")

data.head()
```

```python
data["measure"].unique()
```

```python
conc_avg = data.number[data.measure == "concentration"].mean()
dep_avg = data.number[data.measure == "deposition"].mean()
print(f"Average atmospheric microplastics composition = {conc_avg:.4f} particles/m3")
print(f"Average microplastics deposition rate = {dep_avg:.0f} particles/m2/d")
```
