## Analyse of the data stored in obs_evangelou2026-revised.csv

Data retrived on 2026-11-18 from https://zenodo.org/records/23046811 see download_microplastics_data.py

1. Plot the atmospheric concentration of microplastics as a function of year
2. Plot showing atmospheric deposition rates as a function of year


```python
#imports

from pandas import read_csv
import matplotlib.pyplot as plt

data = read_csv("obs_evangelou2026-revised.csv")
```

```python
#Amospheric concentration
plt.plot(data.year[data.measure =="concentration"], data.number[data.measure =="concentration"], "o")
plt.xlabel('Year')
plt.ylabel('Atmospheric Concentration of Microplastics particle/m3')
```

```python
#Atmospheric Deposition
plt.plot(data.year[data.measure =="deposition"], data.number[data.measure =="deposition"], "o")
plt.xlabel('Year')
plt.ylabel('Atmospheric Deposition of Microplastics particle/m3/day')
```
